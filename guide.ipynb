# Guide

This will be a manual including information on ISO-639-3, ISO-15924, and column names.

#### Dataset Item Dictionary

| Table | Column | Format | Meaning |
| ----- | ------ | ------ | ------- |
| Sign | sign_id | S#### | Unique identifier for the sign |
| Observation | observation_id | O#### | Unique identifier for the observation |
| Observation | date_observed | mm.dd.yyyy | Date of the observation |
| Observation | latitude | WGS84 | Sign location latitude at the time of observation |
| Observation | longitude | WGS84 | Sign location longitude at the time of observation |
| Observation | status | Text | The status of the sign (is it present, removed, moved) |
| Observation | condition | Text | The condition of the sign (clean, dirty, etc.) |
| Photo | photo_id | P#### | Unique identifier for the photo |
| Photo | date_captured | mm.dd.yyyy | Date the photo was taken |
| Photo | filename | Text | The file name of the image |
| Photo | photographer | Text | The name of the photographer |
| All | notes | Text | Additional comments |
