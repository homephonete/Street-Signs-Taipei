# Search the Data

The processed dataset complete with the sign contents extracted by the OCR is very big, so I wanted to make it searchable, like a corpus object. To achieve this, I used [ITables](https://itables.org/) and my own combined dataset of my field surveys in Tainan and Wulai.

```python
import pandas as pd

excel_path = "../data/spreadsheets/street_signs_taipei_sheets.xlsx"

observations = pd.read_excel(
    excel_path,
    sheet_name="OBSERVATION"
)

search_df = observations[
    [
        "sign_id",
        "observation_id",
        "survey_id",
        "observation_date",
        "sign_latitude",
        "sign_longitude",
        "sign_content",
        "sign_status",
        "sign_condition",
        "sign_photo_filename",
        "observation_notes"
    ]
].copy()

search_df["observation_date"] = pd.to_datetime(
    search_df["observation_date"],
    errors="coerce"
)

search_df["month"] = (
    search_df["observation_date"]
    .dt.to_period("M")
    .astype(str)
)

from itables import show

display_df = search_df[
    [
        "sign_id",
        "observation_id",
        "survey_id",
        "month",
        "sign_content",
        "sign_status",
        "sign_condition",
        "sign_photo_filename"
    ]
]

display_df["sign_content"] = (
    display_df["sign_content"]
    .fillna("")
    .str.slice(0, 100)
)

show(
    display_df,
    paging=True,
    columnDefs=[
        {
            "targets": "_all",
            "className": "dt-left"
        }
    ]
)
```
