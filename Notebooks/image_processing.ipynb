import os
import gc
from pathlib import Path
import pandas as pd

os.environ["FLAGS_enable_pir_api"] = "0"

from paddleocr import PaddleOCR

input_dir = Path(
    r"C:/Users/omistaja/Desktop/AS Project/Data/Photos/S0001b"
)

output_excel = (
    r"C:\Users\omistaja\Desktop\AS Project\Data\ocr_results4.xlsx"
)

ocr = PaddleOCR(
    enable_mkldnn=False,
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    use_textline_orientation=False
)

extensions = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}

files = [
    f for f in input_dir.iterdir()
    if f.suffix.lower() in extensions
]

rows = []

for i, file in enumerate(files, start=1):

    print(f"{i}/{len(files)}: {file.name}")

    try:
        results = ocr.predict(
            input=str(file),
            text_det_limit_type="max",
            text_det_limit_side_len=1280
        )

        # Collect all recognized text from this image
        text_lines = []

        for result in results:
            data = result.json

            # PaddleOCR 3.x
            rec_texts = data["res"].get("rec_texts", [])
            text_lines.extend(rec_texts)

        rows.append({
            "filename": file.name,
            "text": "\n".join(text_lines)
        })

        del results
        gc.collect()

    except Exception as e:

        print(f"FAILED: {file.name}")

        rows.append({
            "filename": file.name,
            "text": "",
            "error": str(e)
        })

df = pd.DataFrame(rows)

# Save to Excel
df.to_excel(output_excel, index=False)

print("Finished!")
print(output_excel)
