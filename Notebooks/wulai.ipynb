# Selected signs: Wulai Ulay 烏來

![Wulai Waterfall 烏來瀑布](IMG_0758.jpg)

In order to try out the workflow, I conducted a field survey in Wulai. I selected Wulai as my site due to its multilingual character shaped by a history of colonialism and subsequent tourism. The Wulai Village has been experiencing steady interest by tourists since the 1960s, and tourism has become a significant source of income for the local Atayal community (Hitchcock, 2003; Yoshimura & Wall, 2010). 

![Survey route](S0002.png)

This data set offers a glimpse into the current linguistic landscape of Wulai. It consists of 434 signs and their associated metadata. Below is a demonstration of breaking down OCR output into individual phrases and identifying languages and scripts.
