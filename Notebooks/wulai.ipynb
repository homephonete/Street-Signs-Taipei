# Example Survey: Wulai Ulay 烏來

![Wulai Waterfall 烏來瀑布](IMG_0758.JPG)

In order to try out the workflow, I conducted a field survey in Wulai. I selected Wulai as my site due to its multilingual character shaped by a history of colonialism and subsequent tourism. The Wulai Village experienced a steady interest by tourists beginning in the 1960s. Despite the traffic of specifically gloabl tourists slowing down since, tourism remains a significant source of income for the local Atayal community (Hitchcock, 2010; Yoshimura & Wall, 2010). This creates what Pratt (1991) termed as a contact zone; a place where cultures interact under certain conditions of power. This makes Wulai a compelling site to study language distribution in street signs: what indexes can be found and what do they reveal about power relations between language groups.

![Survey route](S0002.png)

This data set offers a glimpse into the current linguistic landscape of Wulai. It consists of 434 signs and their associated metadata. Below is a demonstration of breaking down OCR output into individual phrases and identifying languages and scripts.

The next step would be compiling a new corpus object of these text elements using previous steps included on this site. A comprehensive corpus of text elements found on a site, together with a database of images and metadata about independent signs through time, can help us understand what language are given space within a location; which languages are present, which are not, what kind of signs in which languages go through what kinds of changes. Which languages tend to appear on official government-maintained signage such as street names? How are non-latin texts romanized (for instance; in this data set you can find the discrepancy between the standard English romanization of Wulai and the Atayal name Ulay).
