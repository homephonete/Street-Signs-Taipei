# Introduction

Street Signs Taipei is an experiment into the documentation of street signs in an effort to understand and research living multilingualism in our environment. The purpose of this site is to explore the factors that need to be accounted for when documenting street signs through a reproducible workflow, demonstrated with a compact data set.

## Why study street signs?

Street signs are indexical; their meanings are based on their material location and social context and help us analyze the relationship between the two in a given place . In terms of cultural landscape documentation, street signs therefore offer insight into how people assign meaning within a space. This project is rooted in the theoretical framework by Scollon and Scollon Wong (2003) which outlines the study of discourse in space, or *geosemiotics*.

According to Scollon and Scollon (2003), a sign is any material object that refers to something other than itself. This project adopts an interpretation of this definition which is limited to any material object that refers to something other than itself by means of any natural language, meaning that this project is especially interested in the use and distribution of language in space.

Rooted in this idea of capturing language in space, this site first introduces the methodology pipeline from collecting data to processing and displaying it. It then demonstrates the use of this pipeline to collect, explore, and begin to analyze linguistic features of selected sites.
