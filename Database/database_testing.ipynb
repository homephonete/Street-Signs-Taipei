```python
import sqlite3
import pandas as pd
df=pd.read_excel("C:/Users/omistaja/Desktop/AS Project/Data/Spreadsheets/street_signs_taipei_sheets.xlsx")

conn = sqlite3.connect("taipei_street_signs_sqlite_practice.sqlite")

cur = conn.cursor()

cur.execute("""CREATE TABLE IF NOT EXISTS SIGN(
    sign_id TEXT, 
    sign_location TEXT, 
    sign_notes TEXT
    )"""
              )

df.to_sql('SIGN',conn,if_exists='append', index=False)

pd.read_sql("SELECT sign_location FROM SIGN",conn)
```
